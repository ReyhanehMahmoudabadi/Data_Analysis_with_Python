# List to Numpy Array :

```python
import numpy as np
```

```python
numbers=[[1,5,3],[5,22,6]]

x=np.array(numbers)
print(x)
```

```python
print(x+2)
```

```python
print(x/5)
```

```python
print(x.dtype)
```

### Numpy Data Type Conversion

```python
f=np.array(numbers,dtype=np.int8)
print(f)
print(f.dtype)
```
