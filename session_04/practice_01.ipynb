```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset("titanic")
df
```

```python
df.info()
```

```python
plt.hist(df["survived"],bins=2)
```

```python
plt.figure(figsize=(10,3))

plt.subplot(121)
plt.title("Man")
plt.hist(df[df["who"]=="man"]["survived"],bins=2)

plt.subplot(122)
plt.title("Woman")
plt.hist(df[df["who"]=="woman"]["survived"],bins=2)
```
